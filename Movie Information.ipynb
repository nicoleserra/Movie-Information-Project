API stands for Application Programming Interface and REST stands for Representational State Transfer.
When REST API is requested by a customer a representation of the state of the resource is transferred to the 
endpoint or requester. This information can be delivered in several formats. For example HTML, XLT, PHP, Python, plain text or JSON, which is the most commonly used. Clients initiate all the interactions as requests cannot be done by servers and clients cannot respond. The interaction between a program and a service through a REST API usually involves resources, which are the most important entities or data objects that can be manipulated. Unique URLs are used to identify resources. It also involves HTTP methods to perform actions on the resources, such as GET, POST, PUT and DELETE. The program sends an HTTP request to the service's API endpoint, providing any necessary data or parameters and specifying the preferred HTTP method. The service processes the request and produces an HTTP response after receiving it. The reaction includes a status code showing either the success or failure of the request.

Overview:

API stands for application programming interface. It is a piece of software that acts as a link between two applications. Any applications that make use of APIs communicate with a server by connecting to the Internet. The server then recovers that information, deciphers it, plays out the vital activities and sends it back to your phone. The application then deciphers that information and presents you with the data you need in a meaningful manner. The OMDB Programming interface is a web administration that gives admittance to an immense data set of film related data. It permits designers to recover film information automatically by sending HTTP requests to the OMDB Programming interface server. You interact with the OMDB API by first getting a programming interface key. The programming interface key is an identifier for your application and permits the programming interface server to follow your use. There are a number of endpoints defined by the OMDB API, each suited to a distinct use case or type of data. To interact with these endpoints, HTTP methods like GET, POST, and occasionally PUT and DELETE are utilized. The API allows you to pass query parameters in the API request URL to specify the details of the data you want to retrieve. The OMDB API may enforce rate limits. Rate limits define the highest number of requests you can make inside a particular time span. Typically, a programming language or a tool that can make HTTP requests, like Python's requests library, is used to interact with the OMDB API.
